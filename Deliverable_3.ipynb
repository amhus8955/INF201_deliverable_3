```python
import random

from battle import Battle
from character import Character, Mage, Warrior
from dice import Dice
from items import Weapon

random.seed(42)  # reproducible results


def present(obj) -> None:
    """Show the attributes and public methods of an object's class."""
    cls = type(obj)
    print(f"{cls.__name__} attributes: {vars(obj)}")
    print(f"{cls.__name__} methods: {[n for n in dir(cls) if not n.startswith('_')]}")
```

## The classes

```python
sword = Weapon("Sword", Dice(sides=8))
present(Dice(sides=20))
present(sword)
present(Character("Hero"))
present(Warrior("Thorin"))
present(Mage("Gandalf"))
present(Battle(Warrior("Thorin"), Mage("Gandalf")))
```

## Battles

```python
Battle(Warrior("Thorin"), Mage("Gandalf")).simulate_battle();
```

```python
axe = Weapon("Axe", Dice(sides=8))
staff = Weapon("Staff", Dice(sides=6))
Battle(Warrior("Thorin", weapon=axe), Mage("Gandalf", weapon=staff)).simulate_battle();
```

```python
Battle(Character("Hero", weapon=sword), Character("Goblin", hp_max=60, damage=8)).simulate_battle();
```

## Discussion

- **Encapsulation:** `_hp` is only changed by the character itself, through `take_damage()` and `heal()`. `attack()` calls the target's `take_damage()`, and `hp` and `is_alive` are read-only properties.
- **Composition:** A `Weapon` has a `Dice`, and a `Character` has a `Weapon`. A character without a weapon uses its default `damage`.
- **Inheritance:** `Warrior` overrides `take_damage()` to subtract armor. `Mage` overrides `attack()` to spend 10 mana on a double-damage spell, and otherwise attacks normally and restores 5 mana.
- **Battle:** `play_round()` lets the hero attack first, and the enemy only attacks back if it is still alive. `simulate_battle()` repeats this until one character dies.
- **Randomness:** The results vary with the dice rolls. We use `random.seed(42)` so the notebook gives the same output each time.
